```python
import numpy as np
import pandas as pd
import nltk
import string as s
import re
from nltk.corpus import stopwords
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
from wordcloud import WordCloud,STOPWORDS

from sklearn.feature_extraction.text  import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics  import f1_score,accuracy_score
from sklearn.metrics import  confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import SGDClassifier
from lightgbm import LGBMClassifier
```

```python
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
train_df=pd.read_csv("train.csv",header=0,names=['classid','title','desc'])
test_df=pd.read_csv("test.csv",header=0,names=['classid','title','desc'])
```

```python
train_df.head()
```

```python
test_df.head()
```

```python
train_df['text'] = train_df['title'] + " " + train_df['desc']
test_df['text'] = test_df['title'] + " " + test_df['desc']
```

```python
train_x = train_df['text']
train_y = train_df['classid']

test_x = test_df['text']
test_y = test_df['classid']
```

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = str(text).lower()

    # remove special characters
    text = re.sub(r'[^a-z\s]', '', text)

    words = text.split()

    # remove stopwords
    words = [word for word in words if word not in stop_words]

    return " ".join(words)
```

```python
train_x = train_x.apply(preprocess)
test_x = test_x.apply(preprocess)

print(train_x.head())  # MUST look like normal words
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000)  # limit features (better performance)

train_vec = tfidf.fit_transform(train_x)
test_vec = tfidf.transform(test_x)
```

```python
sns.countplot(x=train_y)
plt.show()
```

```python
tfidf = TfidfVectorizer(min_df=3)

train_vec = tfidf.fit_transform(train_x)
test_vec = tfidf.transform(test_x)

print("Features:", len(tfidf.get_feature_names_out()))
```

```python
train_arr = train_vec.toarray()
test_arr = test_vec.toarray()
```

```python
from sklearn.naive_bayes import MultinomialNB

NB = MultinomialNB(alpha=0.16)
NB.fit(train_arr, train_y)

pred = NB.predict(test_arr)
```

```python
from sklearn.tree import DecisionTreeClassifier

DT = DecisionTreeClassifier()
DT.fit(train_arr, train_y)

pred = DT.predict(test_arr)
```

```python
from sklearn.naive_bayes import GaussianNB

GNB = GaussianNB()
GNB.fit(train_arr, train_y)

pred = GNB.predict(test_arr)
```

```python
from sklearn.linear_model import SGDClassifier

SGD = SGDClassifier(early_stopping=True)
SGD.fit(train_arr, train_y)

pred = SGD.predict(test_arr)
```

```python
!pip install lightgbm

from lightgbm import LGBMClassifier

lgbm = LGBMClassifier()
lgbm.fit(train_arr, train_y)

pred = lgbm.predict(test_arr)
```

```python
def eval_model(y, y_pred):
    print("F1 Score:", f1_score(y, y_pred, average='micro'))
    print("Accuracy:", accuracy_score(y, y_pred))
    print("Accuracy %:", round(accuracy_score(y, y_pred)*100, 2))
```

```python
eval_model(test_y, pred)
```

```python
def confusion_mat():
    cm = confusion_matrix(test_y, pred)

    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

confusion_mat()
```
